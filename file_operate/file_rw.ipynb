## python 文件操作基础

```python
#文件写入
file = open('H:/learing/Python/py_tourist/file_operate/note.txt', 'w')
file.write('this one line note content.')
file.close()
```

```python
#文件读取
file = open('H:/learing/Python/py_tourist/file_operate/note.txt', 'r')
content = file.read()
print(content)
file.close()
```

```python
# 读取文件 cursor
file = open("H:/learing/Python/py_tourist/file_operate/note.txt", "r")
cursor = file.tell()
file.close()
print(cursor)

#偏移读取数据
file = open("H:/learing/Python/py_tourist/file_operate/note.txt", "r")
cursor0 = file.tell()
file.read(5)
cursor = file.tell()
print(cursor)
# param1:偏移量 param2:起始位置
file.seek(3,0)
seek_cursor = file.tell()
print(seek_cursor)
file.close()
```

```python
# 按行读取数据
file = open("H:/learing/Python/py_tourist/file_operate/note_lines.txt", "r")
print(file.readline())
#从头读
file.seek(0)
lines = file.readlines()
for line in lines:
    print(line)
    print("+++")
file.close()
```

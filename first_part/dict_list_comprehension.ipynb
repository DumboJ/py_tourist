# python 中的字典和数组推导式

```python
#求 0-10 所有偶数的平方
#普通写法
nums_list =[]
for num in range(1,11):
    if num % 2 == 0:
        nums_list.append(num**2)
print(nums_list)

# 列表推导式
dict_nums = [i**2 for i in range(1,11) if i%2==0]
print(dict_nums)

#字典推导式
new_dict = {}
dicts = {"name":"Lucy","age":25,"location":"beijing"}
for i in dicts:
    new_dict[i] = 0
print(new_dict)

comprehension_dict = {}
comprehension_dict = {i:0 for i in dicts}
print(comprehension_dict)
```

# python 中的字典和数组推导式

```python
#求 0-10 所有偶数的平方
#普通写法
nums_list =[]
for num in range(1,11):
    if num % 2 == 0:
        nums_list.append(num**2)
print(nums_list)

# 列表推导式
dict_nums = [i**2 for i in range(1,11) if i%2==0]
print(dict_nums)

# map lambda 表达式写法
nums_map_lambda = map(lambda n: n**2 , range(1,11))
print(list(nums_map_lambda))

#字典推导式
new_dict = {}
dicts = {"name":"Lucy","age":25,"location":"beijing"}
for i in dicts:
    new_dict[i] = 0
print(new_dict)

comprehension_dict = {}
comprehension_dict = {i:0 for i in dicts}
print(comprehension_dict)
```

```python
# 找出 1-11的偶数
#推导式
even_nums = [n for n in range(1,11) if n%2==0]
print(even_nums)

#lambda 写法
filter_even_nums = filter(lambda n: n%2==0,range(1,11))
print(list(filter_even_nums))
```

```python
# I want a (letter,num) pair for each letter in 'abcd' and each number in 1234
for_num = []
for letter in 'abcd':
    for num in range(1,5):
        for_num.append((letter,num))
print(for_num)

#推导式
comprehension_letter_num = [(letter,num) for letter in 'abcd' for num in range(1,5)]
print(comprehension_letter_num)
```

```python
# dict comprehension
#四大名著和作者对应的dict
book_name = ['三国演义','水浒传','西游记','红楼梦']
author = ['罗贯中','施耐庵','吴承恩','曹雪芹']

# for loop
# res_dict = {}
# for book_name, author in zip(book_name, author):
#     res_dict[book_name] = author
# print(res_dict)

#推导式
comprehension_dict = {book_name: author for book_name, author in zip(book_name, author)}
print(comprehension_dict)

#剔除施耐庵的作品
filter_res_dict = {book_name: author for book_name, author in zip(book_name, author) if author != '施耐庵'}
print (filter_res_dict)
```

```python
# Set 的推导式
nums = [1,1,2,3,5,3,4,6,8,0,10,9,10]
nums_set = set()
for num in nums:
    nums_set.add(num)
print(nums_set)

#comprehension expression
com_set = {num for num in nums_set}
print(com_set)
```

```python
#计算数族元素的平方
nums = [1, 2, 3, 4, 5, 6, 8, 9, 10]
def generate_num(nums):
    for n in nums:
        yield n*n
for_res = generate_num(nums)
print(list(for_res))

#推导式
com_res = (n*n for n in nums)
print(list(com_res))
```

#### 关于Python编程中的控制流

##### 1. if语句

```python
x = int(input('Please enter a number: '))
if x < 0 :
    x = 0
    print('navigate change to zero')
elif x == 0:
    print('Zero')
elif x == 1:
    print('Single')
else:
    print('More')
```

2.for 循环，不是迭代数值，而是元素顺序

```python
words = ['cat', 'window', 'defenestrate']
for word in words:
    print(word,len(word))
```

难以迭代多项集操作，策略1：复制副本 2.新建多项集

```python
users = {'Hans': 'active', 'Éléonore': 'inactive', '景太郎': 'active'}
for user,state in users.copy().items():
    if state == 'active':
        del users[user]
print(users)
```

```python
result_user = {}
users = {'Hans': 'active', 'Éléonore': 'inactive', '景太郎': 'active'}
for user,state in users.items():
    if state == 'active':
        result_user[user] = users[user]
print(result_user)
```

3.range 函数生成等差数列

```python
for i in range(5):
    print(i,end=',')
```

```python
#生成的序列不会包含给定的终止值，而且可以正反向给定步长递增
print(list(range(10)))

print(list(range(0,10,3)))
print(list(range(-10,-30,-3)))

## 按照索引迭代
a = ['Mary', 'had', 'a', 'little', 'lamb']
for i in range(len(a)):
    print(i,a[i])

#内置函数会直接计算range的值
print(sum(range(5))) #0+1+2+3+4
```

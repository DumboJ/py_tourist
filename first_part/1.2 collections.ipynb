### 集合的一些常用操作

```python
cities = ['beijing','chengdu','chongqing','kunming']
province = ['beijing','sichuan','chongqing','yunnan']
nums = [1,2,3,4,5]
print(cities)

#append() 结尾添加元素
cities.append('hangzhou')
print(cities)

# insert() 某个位置下标添加元素
cities.insert(0,province)
print(cities)

#extend 如果是添加集合，会平铺元素到同一层级
province.extend(cities)
print(province)

#reverse() 翻转
province.reverse()
print(province)

#sorted()
print(sorted(nums))
```

### 对于数组的 join/split

```python
names = ['Bob','Lucy','Tom']
joinNames = ' - '.join(names)
print(joinNames)

splitList = joinNames.split('-')
print(splitList)
```

# immutable: tuple
# mutable: list

```python
# mutable :list case
course = ['math','physics','science']
course1 = course
course[0] = 'art'
print(course)
print(course1)

#immutable: tuple case
tuple_1 = ('math','physics','science')
tuple_2 = tuple_1
tuple_1[0] = 'art' # TypeError: 'tuple' object does not support item assignment
print(tuple_1)
print(tuple_2)
```

# Sets 无顺序集合：每次打印元素位置不一样

```python
#无序，集合元素唯一性
set_1 = {'1','2','3','4','5','2'}
print(set_1)
print('1' in set_1) #效率比集合或者tuple高

#求交集
set_2 = {'1','2','9','6','8'}
print(set_2.intersection(set_1))

#求差异
print(set_2.difference(set_1))

#求并集
print(set_1.union(set_2))
```

# 字典 类似于Java-HashMap

```python
#定义
student = {'name':'John','age':25,'course':['Math','Physics','Science']}
print(student)

#获取元素 get(key)/index
name = student.get('name')
print(name)
print(student['name'])
print(student.get('phone','Notfound'))
student['phone'] = 133737128319
print(student.get('phone','Notfound'))

#增改元素
student['phone'] = 15289120932
student.update({'name':'Lucy','age':28})
print(student)

#删除元素 del pop（删除并获取元素值）
del student['phone']
print(student)
student.update({'phone':133737128319})
print(student.pop('phone'))
print(student)

#遍历元素集合
for key ,value in student.items():
    print(key,value)

for key in student:
    print(key)
print(len(student))
print(student.keys())
print(student.values())
```

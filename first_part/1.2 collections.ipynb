1.List 可变集合，不限制集合元素数据类型

```python
intList = [1,2,3,4,5]
#也可以掺杂其它的数值类型
multiList = [1,2,'str']
```

2.索引访问和切片、合并

```python
intList[0]
intList + [89,10]
```

3.添加元素

```python
intList.append(99)
print(intList)
```

4.简单赋值不会复制数据

```python
namedList = intList
print(namedList)
```

5.切片操作会返回列表的浅拷贝

```python
sliceList  = namedList[:3]
print(sliceList)
print(namedList)
```

6.其它操作不再一一展示查看官方文档：赋值、切片、二维数组

7.输出斐波那契

```python
a,b = 0,1
while a < 10 :
    print(a)
    a,b = b,a+b
```

8.关键字 end 可以取消换行，并用另一个字符出汗结尾

```python
a,b = 0,1
while a < 10 :
    print(a,end=',')
    a,b = b,a+b
```

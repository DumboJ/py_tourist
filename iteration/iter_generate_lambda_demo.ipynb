## 函数生成器与迭代器 lambda 表达式

```python
#函数生成器
#以步长为0.5增加，从特定值到结束的数据
def step_incr(start,end,step):
    res = []
    while start < end:
        yield start
        start+=step
        res.append(start)
    return res
print(list(step_incr(1,10,0.5)))
```

```python
#函数迭代器
nums = [1,2,3,4,5]
it = iter(nums)
print(next(it))
print(next(it))
print(next(it))
```

```python
#lambda 表达式
def add(x,y):
    return x+y
print(add(1,2))

#lambda 写法
print(lambda x,y:x*y)
addl = lambda x,y:x+y
print(type(addl))
print(addl(1,2))
```

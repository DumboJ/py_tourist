### Simple function

```python
def ask(question,answer='No response'):
    return '{} \n {}'.format(question,answer)
print(ask('What\'s your name?','Lucy'))
print(ask('Do you know who am I?'))
```

### 解包

```python
#输出可变参数时自动区分
def student_info(*args,**kwargs):
    print(args)
    print(kwargs)
student_info('Math','Art',name='Lily',age=23,favorite_color='blue')

courses = ['Python','C++','Java']
basic_infos = {'name':'Lucy','age':23,'favorite_color':'blue' }

#非解包写法，不会识别打印
student_info(courses,basic_infos)
#(['Python', 'C++', 'Java'], {'name': 'Lucy', 'age': 23, 'favorite_color': 'blue'})
#{}

#解包写法，正常打印
student_info(*courses,**basic_infos)
# ('Python', 'C++', 'Java')
# {'name': 'Lucy', 'age': 23, 'favorite_color': 'blue'}
```

### 练习：输入年份和月份，判断月份天数

```python
#多添加一位0保证取元素顺序正确
month_days = [0,31,28,31,30,31,30,31,31,30,31,30,31]
#是否为闰年
def is_leap(year):
    return year%4==0 and (year%100!=0 or year%400==0)

def print_month_days(year,month):
    if not 1<= month <= 12:
        return 'Invalid month'

    if is_leap(year) and month == 2:
        return '29'

    return month_days[month]

print(print_month_days(1992,1))
print(print_month_days(2100,2))
```

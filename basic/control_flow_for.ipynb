### for 循环

```python
### 九九乘法表
for x in range(1,10):
    for y in range(1,x+1):
        print(f'{x}*{y}={x*y}')
```

### for loop with continue & break

```python
#with break
for x in range(1,10):
    if x == 3 :
        print("Found 3")
        break
    print(x,end = ' ')

#with continue
for x in range(1,10):
    if x == 3 :
        print("Found 3",end = ' ')
        continue
    print(x,end = ' ')
```

### for loop with range

```python
for i in range(1,10):
    print(i,end = ' ')
```

### two floor for loop

```python
args = ['a','b','c','d','e']
for x in range(1,6):
    for y in args:
        print(x,y,end = ' ')
```

### while loop

```python
# while break loop
top = 0
while top < 10:
    if top == 6:
        break
    print(top,end = ' ')
    top += 1
```

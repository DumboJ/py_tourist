### if condition

```python
language = 'Golang'
if language == 'Python':
    print("This language is Python")
elif language == 'Java':
    print("This language is Java")
else :
    print('No Match')
```

### and / or / not

```python
username = 'admin'
authenticated = True
if not authenticated:
    print('Please login first')
else:
    print('Welcome Authenticated')

if username == 'admin' and authenticated:
    print('Welcome Admin')
elif username != 'admin' or not authenticated:
    print('please login first')
else:
    print('Unknown User')
```

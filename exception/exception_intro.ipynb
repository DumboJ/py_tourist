## Python的异常处理

```python
from warnings import catch_warnings

#异常示例
try:
    deliver = 5/0
except ZeroDivisionError as e:
    print('除数不能为0 %s',e)
```

```python
#自定义异常
try:
    raise ValueError('<UNK>')
except ValueError as e:
    print('ValueError',e)
```

```python
#try catch finally
#1.错误写法，读取失败时 a 无法正常赋值
try:
    a = open("name1.txt")
except FileNotFoundError as e:
    print('FileNotFoundError',e)
finally:
    a.close()
```

```python
#try catch finally
#2.使用 with
try:
    with open("name1.txt") as f:
        pass
except FileNotFoundError as e:
    print('FileNotFoundError',e)
```

```python
#try catch finally
#3.提前定义
a = None
try:
    open("name1.txt")
except FileNotFoundError as e:
    print('FileNotFoundError',e)
finally:
    if a is not None:
       a.close()
```
